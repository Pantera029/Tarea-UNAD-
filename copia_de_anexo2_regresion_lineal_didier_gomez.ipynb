```python
# ============================================================
# ANEXO 2 - REGRESIÓN LINEAL SIMPLE
# Curso: Estadística y Probabilidad - Fase 2
# Estudiante: Didier Adrian Gomez Ortiz
# Variables: Horas de Experiencia Laboral (X) vs Salario Esperado (Y)
# ============================================================

# 1. Importar librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
from scipy.stats import pearsonr
from google.colab import files

print("✅ Librerías importadas correctamente")

# 2. Subir la base de datos
print("\nPor favor selecciona tu archivo Excel (la plantilla con la hoja 'Base Datos')")
uploaded = files.upload()

nombre_archivo = list(uploaded.keys())[0]
df = pd.read_excel(nombre_archivo, sheet_name='Base Datos')

print(f"\n✅ Base de datos cargada: {nombre_archivo}")
print(f"Total de registros: {len(df)}")
print(df.head())

# 3. Seleccionar variables
X = df['Horas de Experiencia Laboral'].values.reshape(-1, 1)
Y = df['Salario Esperado'].values

print("\nVariable independiente (X): Horas de Experiencia Laboral")
print("Variable dependiente (Y): Salario Esperado")
print(f"Cantidad de datos utilizados: {len(X)}")

# 4. Crear y entrenar el modelo
modelo = LinearRegression()
modelo.fit(X, Y)

pendiente = modelo.coef_[0]
intercepto = modelo.intercept_
Y_pred = modelo.predict(X)

print("\n" + "="*65)
print("ECUACIÓN DE LA REGRESIÓN LINEAL")
print("="*65)
print(f"Y = {intercepto:,.2f} + {pendiente:,.2f} × X")
print(f"\nInterpretación:")
print(f"• Por cada hora adicional de experiencia laboral, el salario esperado aumenta en promedio ${pendiente:,.2f}")
print(f"• Cuando X = 0 (sin experiencia), el salario base esperado es aproximadamente ${intercepto:,.2f}")

# 5. Calcular r, R² y MAE
r, p_value = pearsonr(X.flatten(), Y)
r2 = r2_score(Y, Y_pred)
mae = mean_absolute_error(Y, Y_pred)

print("\n" + "="*65)
print("MEDIDAS DE CORRELACIÓN Y AJUSTE DEL MODELO")
print("="*65)
print(f"Coeficiente de correlación de Pearson (r): {r:.4f}")
print(f"Coeficiente de determinación (R²):         {r2:.4f}  →  {r2*100:.2f}%")
print(f"Error Absoluto Medio (MAE):               ${mae:,.2f}")
print(f"Valor p:                                  {p_value:.4e}")

# Nivel de correlación según la escala del curso
if abs(r) == 1:
    nivel = "Perfecta"
elif 0.90 < abs(r) < 1:
    nivel = "Excelente"
elif 0.80 < abs(r) < 0.90:
    nivel = "Aceptable"
elif 0.60 < abs(r) < 0.80:
    nivel = "Regular"
elif 0.30 < abs(r) < 0.60:
    nivel = "Mínima"
else:
    nivel = "No hay correlación"

print(f"\nNivel de correlación: {nivel}")
print(f"Tipo de correlación:  {'Positiva' if r > 0 else 'Negativa'}")

# 6. Diagrama de dispersión con línea de regresión
plt.figure(figsize=(12, 7))

plt.scatter(X, Y, alpha=0.6, color='steelblue', edgecolors='white', s=55, label='Datos reales')
plt.plot(X, Y_pred, color='crimson', linewidth=2.5, label=f'Recta de regresión\nY = {intercepto:,.0f} + {pendiente:.2f}X')

plt.title('Diagrama de Dispersión\nHoras de Experiencia Laboral vs Salario Esperado',
          fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Horas de Experiencia Laboral (X)', fontsize=12)
plt.ylabel('Salario Esperado (Y)', fontsize=12)
plt.legend(fontsize=11, loc='upper left')
plt.grid(True, alpha=0.3)

texto = f'r = {r:.3f}\nR² = {r2:.3f} ({r2*100:.1f}%)\nNivel: {nivel}'
plt.text(0.05, 0.95, texto, transform=plt.gca().transAxes, fontsize=12,
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))

plt.tight_layout()
plt.show()

# 7. Tabla de valores reales vs predichos (primeros 15)
comparacion = pd.DataFrame({
    'Horas (X)': X.flatten()[:15],
    'Salario Real (Y)': Y[:15].round(0),
    'Salario Predicho': Y_pred[:15].round(0),
    'Error Absoluto': np.abs(Y[:15] - Y_pred[:15]).round(0)
})

print("\n" + "="*65)
print("TABLA DE VALORES REALES vs PREDICHOS (primeros 15 registros)")
print("="*65)
print(comparacion.to_string(index=False))

# 8. Tres predicciones
valores_prediccion = np.array([200, 500, 800]).reshape(-1, 1)
predicciones = modelo.predict(valores_prediccion)

print("\n" + "="*65)
print("PREDICCIONES CON LA ECUACIÓN DE REGRESIÓN")
print("="*65)
for x_val, pred in zip([200, 500, 800], predicciones):
    print(f"• X = {x_val} horas  →  Salario esperado predicho = ${pred:,.0f}")

# 9. Informe final
print("\n" + "="*70)
print("INFORME FINAL - ANÁLISIS DE REGRESIÓN LINEAL")
print("="*70)
print(f"""
Estudiante: Didier Adrian Gomez Ortiz
Variables:
  - Independiente (X): Horas de Experiencia Laboral
  - Dependiente (Y):   Salario Esperado

Ecuación de regresión:
  Y = {intercepto:,.2f} + {pendiente:,.2f} × X

Resultados principales:
  • Coeficiente de Pearson (r):     {r:.4f}
  • Nivel de correlación:           {nivel} (Positiva)
  • Coeficiente de determinación:   {r2:.4f} ({r2*100:.2f}%)
  • Error Absoluto Medio (MAE):     ${mae:,.2f}

CONCLUSIÓN:
Existe una relación positiva aceptable y estadísticamente significativa
entre las horas de experiencia laboral y el salario esperado
(r = {r:.3f}; R² = {r2*100:.1f}%). El modelo permite predecir de forma
razonable el salario esperado a partir de la experiencia laboral,
reforzando la importancia de este factor en el perfil de empleabilidad
de los estudiantes universitarios.
""")
```
